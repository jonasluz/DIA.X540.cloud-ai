# Azure Speak Service Quickstart
Este notebook demonstra como interagir com o serviço de Fala do Azure AI utilizando o SDK oficial para Python.

```python
import os

from dotenv import load_dotenv
load_dotenv()
```

```python
import azure.cognitiveservices.speech as speechsdk

speech_key = os.getenv("AZURE_SPEECH_KEY")
endpoint = os.getenv("AZURE_SPEECH_ENDPOINT")

speech_config = speechsdk.SpeechConfig(
    subscription=speech_key, 
    endpoint=endpoint)

audio_config = speechsdk.audio.AudioOutputConfig(
    use_default_speaker=True)

# The neural multilingual voice can speak different languages based on the input text.
known_voices = [
    'en-US-Ava:DragonHDLatestNeural', 
    'pt-BR-FranciscaNeural'
]
speech_config.speech_synthesis_voice_name=known_voices[1]

speech_synthesizer = speechsdk.SpeechSynthesizer(
    speech_config=speech_config, 
    audio_config=audio_config)

speech_synthesizer
```

```python
text_to_speak = "Opa, meu fí! Cumé que cê tá? Tudo bem? Ocê quer uma reza para tirar esse mau olhado que tá impregnado nocê?"
```

```python
try:
    speech_synthesis_result = speech_synthesizer.speak_text_async(text_to_speak).get()

    if speech_synthesis_result.reason == speechsdk.ResultReason.SynthesizingAudioCompleted:
        print(f"Speech synthesized for text [{text_to_speak}]")

    elif speech_synthesis_result.reason == speechsdk.ResultReason.Canceled:
        cancellation_details = speech_synthesis_result.cancellation_details
        print(f"Speech synthesis canceled: {cancellation_details.reason}")

        if cancellation_details.reason == speechsdk.CancellationReason.Error:
            if cancellation_details.error_details:
                print(f"Error details: {cancellation_details.error_details}")
                print("Did you set the speech resource key and endpoint values?")
except Exception as e:
    print(f"Error occurred: {e}")
    
```
