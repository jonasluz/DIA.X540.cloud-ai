```python
import sqlite3

import pandas as pd

from speech import GoogleSpeech
```

```python
db = sqlite3.connect("project-2.sqlite")
df_speech = pd.read_sql("SELECT * FROM speech", db)
```

```python
id = 41
target_speech = df_speech.loc[df_speech.id == id]["speech"]
target_speech
```

```python
GoogleSpeech.female_voices
```

```python
for voice in GoogleSpeech.female_voices:
    speaker = GoogleSpeech(voice_name=voice)
    speaker.synthesize(text=target_speech.values[0], 
                       output_filepath=f"output/cancersucks/ID-{id}_{voice}.mp3",
                       prompt="Demonstrate fear and concern in the voice.")
```
