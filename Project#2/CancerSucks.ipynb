```python
import time

from speech import GoogleSpeech
```

```python
talks = [
    "Doutor, boa tarde... estou um pouco ansiosa pelo resultado da minha mamografia. O senhor já tem o laudo?",
    "O que isso quer dizer, doutor? Seria resto de leite?",
    "O senhor disse que vamos precisar fazer uma biópsia? Como é que funciona isso?",
    "Mas, doutor, a gente toma anestesia para esses procedimentos?",
    "Mas doutor, como eu vou escolher se eu nem sei o que cada uma faz?",
    "Então o senhor não pode me dizer o que fazer agora?",
    "Poxa, doutor... eu vou sair daqui com o coração na mão. Estou com medo."
]

prompts = [
    "Demonstrate ansiety and concern while asking about medical test results.",
    "Express confusion and seek clarification about medical exam results.",
    "Inquire about the biopsy procedure with a tone of apprehension.",
    "Ask about anesthesia for medical procedures with a worried tone.",
    "Express uncertainty and ask for guidance on choosing medical options.",
    "Seek advice from the doctor with a tone of desperation.",
    "Convey fear and anxiety about upcoming medical procedures."
]

assert len(talks) == len(prompts)
```

```python
voices = GoogleSpeech.female_voices
len(voices)
```

```python
filename_for = lambda i, voice: f"output/cancersucks/Dialogue-{i+1}_{voice}.mp3"
```

```python
for voice in GoogleSpeech.female_voices:
    for id, (speech, prompt) in enumerate(zip(talks, prompts)):
        speaker = GoogleSpeech(voice_name=voice)
        speaker.synthesize(prompt=prompt, 
                           text=speech, 
                           output_filepath=filename_for(id, voice))

        print(f"Generated: {filename_for(id, voice)}")
        time.sleep(1)  # To avoid hitting rate limits
```
